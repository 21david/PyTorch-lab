```python
import torch
```

### Creating tensors

#### Tensor

```python
t1 = torch.tensor([1, 2, 3])
t1
```

```python
t2 = torch.tensor([[1, 2, 3],
                   [4, 5, 6]])
t2
```

```python
t3 = torch.tensor(
[
  [
    [1,2,3],
    [4,5,6]
  ],
  [
    [10, 11, 12],
    [13, 14, 15]
  ]
])

t3
```

#### Zeroes

```python
z1 = torch.zeros(2, 3)
print(z1)
```

```python
z2 = torch.zeros(size=(2, 3))
print(z2)
```

```python
z3 = torch.zeros(size=(2, 3), dtype=torch.int32)
print(z3)
```

#### Ones

```python
o1 = torch.ones(2, 3, dtype=torch.int32)
o1
```

#### Rand

```python
r1 = torch.rand(2, 3)  # Uniformly distributed from [0, 1)
r1
```

#### Randn

```python
r2 = torch.randn(2, 3) # Normally distributed (bell curve, like human heights), average of 0, standard deviation of 1
r2
```

### Basic operations

```python
t1 = torch.randn(3,4)  
t1
```

```python
t2 = torch.rand(3, 4)  # uniformly distributed (rectangle, like dice rolls)
t2
```

```python
t1 + t2
```

```python
(t1 * 100).round(decimals=1).to(torch.int32)
```
