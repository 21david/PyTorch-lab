### Basic operations

```python
t1 = torch.randn(3,4)  # normally distributed (bell curve, like human heights)
t1
```

```python
t2 = torch.rand(3, 4)  # uniformly distributed (rectangle, like dice rolls)
t2
```

```python
t1 + t2
```

```python
(t1 * 100).round(decimals=1).to(torch.int32)
```
